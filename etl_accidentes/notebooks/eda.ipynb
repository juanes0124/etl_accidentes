```python
# Import libraries for data analysis and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visualization style
sns.set(style="whitegrid")

print("Libraries loaded successfully")
```

```python
# Load the raw dataset
df = pd.read_csv("data/raw/accidentes.csv")

# Show first rows
df.head()
```

```python
# General information about the dataset
df.info()
```

```python
# Statistical summary
df.describe()
```

```python
# Check missing values
missing_values = df.isnull().sum()

missing_values
```

```python
# Visualize missing values
plt.figure(figsize=(10,5))
missing_values.plot(kind="bar")
plt.title("Missing Values per Column")
plt.ylabel("Number of Missing Values")
plt.xlabel("Columns")
plt.show()
```

```python
# Check duplicate rows
duplicates = df.duplicated().sum()

print("Number of duplicated rows:", duplicates)
```

```python
# Accidents by year
accidents_by_year = df['year'].value_counts().sort_index()

plt.figure(figsize=(10,5))
accidents_by_year.plot(kind='bar')
plt.title("Accidents per Year")
plt.xlabel("Year")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
# Top departments with most accidents
top_departments = df['department'].value_counts().head(10)

plt.figure(figsize=(10,6))
top_departments.plot(kind='bar')
plt.title("Top 10 Departments with Most Accidents")
plt.xlabel("Department")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
# Accidents by vehicle type
vehicle_counts = df['vehicle_type'].value_counts()

plt.figure(figsize=(10,6))
vehicle_counts.plot(kind='bar')
plt.title("Accidents by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Number of Accidents")
plt.show()
```

```python
# Distribution of accident severity
severity_counts = df['severity'].value_counts()

plt.figure(figsize=(8,5))
severity_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title("Accident Severity Distribution")
plt.ylabel("")
plt.show()
```

```python
# Correlation matrix (only numeric columns)
corr = df.select_dtypes(include=['int64','float64']).corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
# Boxplot to detect outliers
plt.figure(figsize=(8,5))
sns.boxplot(x=df['number_of_vehicles'])
plt.title("Outlier Detection - Number of Vehicles")
plt.show()
```

```python
print("Key insights from the EDA:")

print("- The dataset contains multiple years of accident records.")
print("- Some departments show a significantly higher number of accidents.")
print("- Certain vehicle types appear more frequently in accident reports.")
print("- There are missing values that need to be addressed during ETL.")
print("- Some numeric variables present potential outliers.")
```
